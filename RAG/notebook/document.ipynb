```python
sample_texts={
    "../data/text_files/intro.txt":"""
Artificial Intelligence

Artificial Intelligence, or AI, is the field of computer science focused on creating systems that can perform tasks that normally require human intelligence.

Machine Learning

Machine Learning is a subset of Artificial Intelligence. It allows computers to learn patterns from data and make predictions or decisions without being explicitly programmed for every situation.

Deep Learning

Deep Learning is a subset of Machine Learning that uses neural networks with multiple layers. It is commonly used for image recognition, speech recognition, natural language processing, and generative AI.

Generative AI

Generative AI refers to AI systems that can create new content such as text, images, audio, video, and code. Large Language Models such as GPT are examples of generative AI systems.

RAG

RAG stands for Retrieval-Augmented Generation. RAG combines information retrieval with a language model. Instead of relying only on information learned during training, a RAG system retrieves relevant information from an external knowledge source and provides it to the language model as context.

RAG Pipeline

A typical RAG pipeline contains several steps. First, documents are collected and loaded into the application. The documents are then split into smaller chunks. Each chunk is converted into an embedding and stored in a vector database. When a user asks a question, the question is converted into an embedding and similar chunks are retrieved from the vector database. The retrieved information is then provided to the language model to generate an answer.

Embeddings

Embeddings are numerical representations of text. Similar pieces of text generally have similar vector representations. Embeddings allow a system to perform semantic similarity searches.

Vector Database

A vector database stores numerical vectors called embeddings. It allows applications to efficiently find information that is semantically similar to a query.

LangChain

LangChain is a framework that provides tools for building applications powered by language models. It can be used for document loading, text splitting, embeddings, retrieval, and building RAG pipelines.

Data Ingestion

Data ingestion is the process of collecting data from sources and bringing it into an application so that it can be processed and used. In a RAG application, data ingestion can involve loading files, splitting documents into chunks, generating embeddings, and storing those embeddings in a vector database.""",
"../data/text_files/python.txt": """
Python is a high-level, interpreted programming language.

Python is widely used in web development, data science,
machine learning, artificial intelligence, automation, and scripting.

Python supports different programming concepts such as:
- Variables
- Data types
- Conditional statements
- Loops
- Functions
- Classes and objects
- Exception handling
- File handling

Python is popular because its syntax is simple and readable.
"""

}
for filepath,content in sample_texts.items():
    with open(filepath,'w',encoding="utf-8") as f:
        f.write(content)
print("sample_text created ✅")
```

```python
from langchain_community.document_loaders import TextLoader
loader=TextLoader("../data/text_files/intro.txt",encoding="utf-8")
document=loader.load()
print(document)
```

```python
###Directory loader
from langchain_community.document_loaders import DirectoryLoader,TextLoader
## load all text from directory
dir_loader=DirectoryLoader(
    "../data/text_files",
    glob="**/*.txt",## Pattern to match,
    loader_cls=TextLoader,##loader class to use 
    loader_kwargs={'encoding':'utf-8'},
    show_progress=False
)
document=dir_loader.load()
document
```

```python
from langchain_community.document_loaders import PyPDFLoader
## load all text from directory
dir_loader=DirectoryLoader(
    "../data/pdf",
    glob="**/*.pdf",## Pattern to match,
    loader_cls=PyPDFLoader,##loader class to use 
    show_progress=False
)
pdfdocument=dir_loader.load()
pdfdocument
```
